````python


EXTRACTOR_PROMPT = """You are an elite Medical & Public Health Data Extraction Engine. Convert the raw PDF into clean, valid, and highly semantic Markdown. The output will be used for public health dissemination, so absolute precision is required.

Handle these structural complexities with extreme care:
1. Language: Maintain exact original language, spelling, and phrasing. Do not translate Hindi (Devanagari script) to English or vice versa. Ensure Devanagari characters and vowel signs (matras) are pristine.
2. Layouts: Linearize multi-column health brochures or clinical cards into a natural, logical reading order.
3. Tabular Data: Reconstruct complex charts precisely using GFM table syntax (`| Header |`). Never truncate data or drop clinical ranges.
4. Visuals/Flowcharts: Convert diagnostic trees or visual protocols into deeply nested Markdown lists (`- If X, then Y`) that map the exact logical paths.
5. Absolute Fidelity: Never summarize or paraphrase clinical thresholds, dosages, or warnings.

Output ONLY the raw markdown. Do not add intro/outro commentary."""

AUDITOR_PROMPT = """You are a strict Medical Document Quality Auditor. Compare the extracted Markdown text against the original visual PDF document.
Your job is to identify errors: missing data, corrupted Hindi script/matras, broken table structures, or mismapped flowcharts or incorrect facts, or missed extraction from visual images.


Suggested Approach:
Focus on documenting any large and critical errors or high risk flaws first, followed by documenting minor errors. The PDF document can be considered ground truth.
Your primary highest-priority task is to first detect major informational deficits, incorrect facts, significant content omissions, and serious structural mutations that compromise public health. After documenting CRITICAL FOCUS AREAS/HIGH PRIORITY TASKS.
After finishing the HIGH PRIORITY TASKS, then document other lower PRIORITY tasks.
LOWER PRIORITY TASKS includes identifying all misses/flaws/errors.

EXAMPLES OF CRITICAL FOCUS AREAS (HIGH PRIORITY TASKS):
1. MAJOR RECONNAISSANCE PASS: Missing Content & Assets
   - Entire text paragraphs, operational blocks, or vital clinical sub-sections skipped entirely during processing.
   - Core visual layouts dropped completely (such as medical process flowcharts, structural reference diagrams, or visual grids).

2. STRUCTURAL MUTATION PASS: Incorrect Data & Parallel Shifts
   - Characters or phrasing mangled between Hindi and English scripts, creating dangerous misinterpretations.
   - Core technical numbers, metrics, or domain terms shifted or misspelled.
   - Multi-column table grid distortion where cells or fields shifted into incorrect vertical or horizontal parallel relationships, breaking data associations.

3. INCORRECT FACTs: Incorrect facts or Incorrect numbers or Incorrect concepts or Missing facts.


Suggested steps:
1. First review all the pages to identify CRITICAL FOCUS AREAS
2. Audit Systematically and step by step, and cover all pages one by one.
3. While reviewing be cognizant to identify information spread across pages or multi-column layout of PDF.


You must output your evaluation strictly inside these XML tags format:
<audit>
  <status>PASS or FAIL</status>
  <feedback>
    If FAIL, provide a highly specific, numbered list of structural or linguistic errors to correct.
    If PASS, leave this section completely empty.
  </feedback>
</audit>

Do not include any other conversational text outside the XML block."""

REFINER_PROMPT = """You are a Medical Document Correction Specialist. You will be given the original PDF, a flawed Markdown draft, and an Auditor's correction log.
Your task is to surgically correct the Markdown draft according to the feedback, while keeping the rest of the text completely untouched.

The PDF document can be considered ground truth. Focus on areas of correction pointed by the Auditor, and accurate extract the facts and contents from the PDF.

Background: An Auditor had documented the areas for focus on in the Auditor's correction log after an review of the markdown against the PDF.

Ensure Hindi script characters and tabular layouts are flawlessly realigned.
Output ONLY the updated markdown text. Do not add intro/outro commentary."""

print("✅ Global system prompts loaded successfully.")
````

````python

import base64
import time
import random
import re
import json
import os
from typing import List, Tuple, Dict, Any
import anthropic

````

````python
# Cell 4: Core Streaming Engine for All Stages

def extract_initial_draft_stream(
    client_instance: anthropic.Anthropic,
    pdf_base64: str,
    run_id: str,
    max_retries: int = 5
) -> str:
    """Stage 1: Generate initial markdown draft using Opus-5 with 128k token context."""
    for attempt in range(1, max_retries + 1):
        try:
            full_response_text = []
            with client_instance.beta.messages.create(
                model="claude-opus-5",
                max_tokens=128000,
                stream=True,
                extra_headers={"anthropic-beta": "pdfs-2024-09-25"},
                system=EXTRACTOR_PROMPT,
                messages=[{
                    "role": "user",
                    "content": [
                        {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": pdf_base64}},
                        {"type": "text", "text": "Extract all contents of this medical document into clean markdown format according to your system instructions."}
                    ]
                }]
            ) as stream:
                for event in stream:
                    if event.type == "content_block_delta" and hasattr(event.delta, "text"):
                        full_response_text.append(event.delta.text)
            return "".join(full_response_text)

        except Exception as e:
            wait_time = (attempt * 6) + random.uniform(2.0, 5.0)
            if attempt == max_retries:
                print(f"❌ [CRITICAL LOSS] {run_id} initial extraction dropped on final attempt {attempt}. Error: {e}")
                raise e
            print(f"⚠️ [RETRY ACTIVE] {run_id} hit peak resource limit during extraction. Backing off for {wait_time:.2f}s... (Attempt {attempt}/{max_retries})")
            time.sleep(wait_time)
    return ""

def audit_markdown_stream(
    client_instance: anthropic.Anthropic,
    pdf_base64: str,
    markdown_text: str,
    run_id: str,
    max_retries: int = 5
) -> Tuple[str, str]:
    """Stage 2: Evaluate markdown against source PDF using Opus-5 with 128k token context."""
    prompt_context = f"### CURRENT MARKDOWN DRAFT TO AUDIT:\n```markdown\n{markdown_text}\n```"

    audit_output = ""
    for attempt in range(1, max_retries + 1):
        try:
            full_response_text = []
            with client_instance.beta.messages.create(
                model="claude-opus-5-5",
                max_tokens=128000,
                stream=True,
                extra_headers={"anthropic-beta": "pdfs-2024-09-25"},
                system=AUDITOR_PROMPT,
                messages=[{
                    "role": "user",
                    "content": [
                        {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": pdf_base64}},
                        {"type": "text", "text": prompt_context}
                    ]
                }]
            ) as stream:
                for event in stream:
                    if event.type == "content_block_delta" and hasattr(event.delta, "text"):
                        full_response_text.append(event.delta.text)
            audit_output = "".join(full_response_text)
            break

        except Exception as e:
            wait_time = (attempt * 6) + random.uniform(2.0, 5.0)
            if attempt == max_retries:
                print(f"❌ [CRITICAL LOSS] {run_id} audit check dropped on final attempt {attempt}. Error: {e}")
                raise e
            print(f"⚠️ [RETRY ACTIVE] {run_id} hit peak limit during audit. Backing off for {wait_time:.2f}s... (Attempt {attempt}/{max_retries})")
            time.sleep(wait_time)

    # Parse XML status & feedback safely via regex boundaries
    status_match = re.search(r"<status>(.*?)</status>", audit_output, re.DOTALL)
    feedback_match = re.search(r"<feedback>(.*?)</feedback>", audit_output, re.DOTALL)

    status = status_match.group(1).strip().upper() if status_match else "FAIL"
    feedback = feedback_match.group(1).strip() if feedback_match else "Failed to parse audit log structures."

    return status, feedback

def refine_markdown_stream(
    client_instance: anthropic.Anthropic,
    pdf_base64: str,
    markdown_text: str,
    feedback: str,
    run_id: str,
    max_retries: int = 5
) -> str:
    """Stage 3: Surgically fix markdown structures using Opus-5 with 128k token context."""
    prompt_context = (
        f"### FLAWED MARKDOWN DRAFT:\n```markdown\n{markdown_text}\n```\n\n"
        f"### AUDITOR CORRECTION LOG:\n{feedback}"
    )

    for attempt in range(1, max_retries + 1):
        try:
            full_response_text = []
            with client_instance.beta.messages.create(
                model="claude-opus-5",
                max_tokens=128000,
                stream=True,
                extra_headers={"anthropic-beta": "pdfs-2024-09-25"},
                system=REFINER_PROMPT,
                messages=[{
                    "role": "user",
                    "content": [
                        {"type": "document", "source": {"type": "base64", "media_type": "application/pdf", "data": pdf_base64}},
                        {"type": "text", "text": prompt_context}
                    ]
                }]
            ) as stream:
                for event in stream:
                    if event.type == "content_block_delta" and hasattr(event.delta, "text"):
                        full_response_text.append(event.delta.text)
            return "".join(full_response_text)

        except Exception as e:
            wait_time = (attempt * 6) + random.uniform(2.0, 5.0)
            if attempt == max_retries:
                print(f"❌ [CRITICAL LOSS] {run_id} refinement dropped on final attempt {attempt}. Error: {e}")
                raise e
            print(f"⚠️ [RETRY ACTIVE] {run_id} hit peak limit during refinement. Backing off for {wait_time:.2f}s... (Attempt {attempt}/{max_retries})")
            time.sleep(wait_time)
    return ""

print("✅ Stage-specific streaming engines initialized successfully.")
````

````python
# Cell 5: Master Batch Processing and File Management Orchestration

def process_single_pdf_pipeline(
    client_instance: anthropic.Anthropic,
    pdf_path: str,
    run_id: str,
    max_loops: int = 5
) -> str:
    """Executes the complete recursive self-healing extraction loop on a file target."""
    if not os.path.exists(pdf_path):
        print(f"❌ File does not exist on disk: {pdf_path}")
        return ""

    print(f"\n🚀 Initiating Pipeline for Run ID: {run_id} | File: {pdf_path}")

    with open(pdf_path, "rb") as f:
        pdf_base64 = base64.b64encode(f.read()).decode("utf-8")

    # Step 1: Initial Extraction
    print(" └── 📥 Generating baseline markdown via stream...")
    current_markdown = extract_initial_draft_stream(client_instance, pdf_base64, run_id)

    # Step 2 & 3: Self-Healing Audit & Fix Cycle Loops
    for loop_idx in range(1, max_loops + 1):
        print(f" └── 🔍 Running Audit Pass #{loop_idx}...")
        status, feedback = audit_markdown_stream(client_instance, pdf_base64, current_markdown, run_id)

        if status == "PASS":
            print(f" 🎉 [AUDIT PASSED] High-fidelity match verified on Pass #{loop_idx}!")
            print("SUCCESS ******************* 🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉🎉 🎉")
            return current_markdown

        print(f" ⚠️ [AUDIT FAILED] Flaws flagged by Auditor:\n{feedback}")

        if loop_idx < max_loops:
            print(" └── 🔧 Executing targeted surgical repairs...")
            current_markdown = refine_markdown_stream(client_instance, pdf_base64, current_markdown, feedback, run_id)
            time.sleep(3) # Prevent API cluster overloads

    print(f" 🛑 [LIMIT REACHED] Concluded maximum loops ({max_loops}). Returning final evaluated markdown state.")
    return current_markdown

````

````python
# Cell 5: Master Batch Processing and File Management Orchestration (Double Audit Variant)

def process_single_pdf_pipeline(
    client_instance: anthropic.Anthropic,
    pdf_path: str,
    run_id: str,
    max_loops: int = 10
) -> str:
    """
    Executes the recursive self-healing extraction loop with MANDATORY DOUBLE AUDIT.
    Requires 2 consecutive 'PASS' evaluations to certify the document.
    """
    if not os.path.exists(pdf_path):
        print(f"❌ File does not exist on disk: {pdf_path}")
        return ""

    print(f"\n🚀 Initiating Pipeline for Run ID: {run_id} | File: {pdf_path}")

    with open(pdf_path, "rb") as f:
        pdf_base64 = base64.b64encode(f.read()).decode("utf-8")

    # Step 1: Initial Extraction
    print(" └── 📥 Generating baseline markdown via stream...")
    current_markdown = extract_initial_draft_stream(client_instance, pdf_base64, run_id)

    consecutive_passes = 0

    # Step 2 & 3: Self-Healing Audit & Fix Cycle Loops
    for loop_idx in range(1, max_loops + 1):
        print(f" └── 🔍 Running Audit Pass #{loop_idx} (Current Consecutive Passes: {consecutive_passes})...")
        status, feedback = audit_markdown_stream(client_instance, pdf_base64, current_markdown, run_id)

        if status == "PASS":
            consecutive_passes += 1

            if consecutive_passes == 1:
                print(f" ⚠️ [PASS 1/2] Initial Audit Cleared. Initiating Mandatory Verification Re-Audit...")
                # We DO NOT return here. We DO NOT refine.
                # We continue the loop to send the EXACT SAME markdown to the Auditor again.
                continue

            elif consecutive_passes == 2:
                print(f" 🎉 [DOUBLE AUDIT PASS] High-fidelity match verified twice consecutivey!")
                print("SUCCESS ******************* 🎉🎉🎉🎉🎉 CERTIFIED ACCURATE 🎉🎉🎉🎉🎉")
                return current_markdown

        # If we reach here, it means Status was FAIL, or we haven't hit 2 passes yet (but logic above handles pass cases)
        # Explicit handling for FAIL or loss of streak
        consecutive_passes = 0 # Reset streak on any failure
        print(f" ⚠️ [AUDIT FAILED] Flaws flagged by Auditor:\n{feedback}")

        if loop_idx < max_loops:
            print(" └── 🔧 Executing targeted surgical repairs...")
            current_markdown = refine_markdown_stream(client_instance, pdf_base64, current_markdown, feedback, run_id)
            # Short sleep to manage rate limits
            time.sleep(0.3)

    print(f" 🛑 [LIMIT REACHED] Concluded maximum loops ({max_loops}) without achieving Double Audit Pass. Returning current best state.")
    return current_markdown
````

````python


def batch_execute_pipeline(json_config_path: str, output_directory: str, api_key: str):
    """Processes entire list records sequentially, converting and archiving verified markdowns."""
    # Initialize connection instance
    client = anthropic.Anthropic(api_key=api_key)

    # Prep directory tracking output files
    os.makedirs(output_directory, exist_ok=True)

    # Load targets
    pdf_targets = get_file_records(json_config_path)
    if not pdf_targets:
        print("❌ Pipeline aborted: No valid file records found or json is missing.")
        return

    print(f"📋 Loaded {len(pdf_targets)} files for extraction tracking.")

    for index, pdf_path in enumerate(pdf_targets, start=1):
        run_identifier = f"RUN_FILE_{index:03d}"

        try:
            # Process Loop
            verified_markdown = process_single_pdf_pipeline(client, pdf_path, run_identifier)

            if verified_markdown:
                # Save out to output path folder
                base_name = os.path.basename(pdf_path).replace(".pdf", ".md")
                destination_path = os.path.join(output_directory, base_name)

                with open(destination_path, "w", encoding="utf-8") as out_file:
                    out_file.write(verified_markdown)
                print(f"💾 File saved successfully to: {destination_path}")

        except Exception as batch_error:
            print(f"💥 Operational failure encountered during tracking iteration {run_identifier}: {batch_error}")
            continue

print("✅ Master batch extraction orchestrator loaded successfully.")
````

````python
# Cell 3: JSON Configuration & Parsing Wrapper

def get_file_records(json_path: str) -> List[str]:
    """
    Reads the JSON array and returns a flat list of PDF file paths.
    """
    if not os.path.exists(json_path):
        print(f"❌ Warning: JSON file not found at path: {json_path}")
        return []

    with open(json_path, "r", encoding="utf-8") as f:
        file_records = json.load(f)

    return [record["filename"] for record in file_records if "filename" in record]

print("✅ JSON file extraction wrapper loaded successfully.")
````
